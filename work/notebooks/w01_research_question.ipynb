## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Trend Direction Value Counts:")
print(f"{df['trend_direction'].value_counts()}\n")

print("Total impressions (90d) for declining pages:")
print(f"{df[df['trend_direction'] == 'down']['impressions_90d'].sum():,.0f}\n")

print("Median days since last update:")
print(df["days_since_last_update"].median())
```

I think data that makes my lane worth for the next 7 weeks are:
1. trend_direction -> shows how many pages are actively declining, which tells us the problem exists at scale and there is a real review queue to build.
2. impressions_90d -> shows how much search visibility is at stake for those declining pages. A declining page with high impressions is a priority; a declining page with near-zero impressions is not.
3. days_since_last_update -> shows how stale the content is, which directly justifies the refresh action. A page that is declining and hasn't been updated in over a year is a stronger candidate than one that was refreshed last month.
